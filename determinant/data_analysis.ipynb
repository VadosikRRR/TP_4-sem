```python
import pandas as pd
import matplotlib.pyplot as plt

PATH_TO_DATA = 'build/result.csv'
```

```python
data = pd.read_csv(PATH_TO_DATA)
data
```

```python
selected_size = 6
selected_deetp = selected_size - 4 if selected_size > 4 else 1
interesting_data = data[(data['Matrix size']==selected_size) & (data['Max deepth']==selected_deetp)].groupby('Thread number')['Time'].median().to_dict()
plt.scatter(interesting_data.keys(), interesting_data.values())
plt.show()
```

```python
max_deepth_data = data[data['Matrix size']==selected_size].groupby('Max deepth')['Time'].median().to_dict()
plt.scatter(max_deepth_data.keys(), max_deepth_data.values())
plt.show()
```
